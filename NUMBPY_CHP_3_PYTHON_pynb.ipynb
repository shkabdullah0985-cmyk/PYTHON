# ***CHAPTER 3*** :- ***ARRAY CREATION ROUTINE***

#WHAT IS ARRAY CREATION ROUTINE ?
**ANSWER**:- ARRAY CREATION ROUTINES IN PYTHON ARE BUILT-IN FUNCTIONS PROVIDED BY THE NUMPY LIBRARY THAT HELP TO CREATE ARRAYS IN DIFFERENT WAYS.

THESE ROUTINES ALLOW US TO CREATE ARRAYS FILLED WITH SPECIFIC VALUES, ZEROS, ONES, RANGES OF NUMBERS, OR EVENLY SPACED VALUES WITHOUT WRITING LONG CODE.

```python
import numpy as np
```

## ***1)EMPTY*** :- IT CREATES AN UNTITELIZED ARRAY OF SPECIFIED SHAPE AND DATA TYPE.

```python
w=np.empty(6)
```

```python
print(w)
```

```python
w.reshape(3,2)
```

```python
p=np.empty([3,2],dtype=int)
```

```python
p
```

## **2)ARANGE**:- IT IS AN ARRAY CREATION ROUTINE IN THE NUMPY LIBRARY THAT IS USED TO CREATE AN ARRAY WITH A SEQUENCE OF NUMBERS.


---



IT STARTS FROM ONE NUMBER AND STOPS BEFORE ANOTHER NUMBER.
YOU CAN ALSO CHOOSE THE GAP (STEP) BETWEEN THE NUMBERS.

```python
v=np.arange(2,21,2).reshape(5,2)
```

```python
v
```

## ***3)ZEROS*** :- IT RETURN A NEW ARRAY OF A SPECIFIED SIZE FILLED WITH ZERO.

```python
q=np.zeros(6)
```

```python
q
```

```python
r=np.zeros([6],dtype=int).reshape(3,2)
```

```python
r
```

## ***4)ONES*** :- IT RETURN THE NEW ARRAY OF SPECIFIED SIZE FILLED WITH ONES.

```python
m=np.ones([4,5],dtype=int)
```

```python
m
```

## ***5)EYE*** :- EYE ROUTINE CAN RETURN THE IDENTITY MATRIX.

```python
l=np.eye(1)
```

```python
l
```

```python
o=np.eye(2)
```

```python
o
```

```python
q=np.eye(3)
```

```python
q
```

## ***6)LINSPACE*** :-THIS FUNCTION IS SIMILAR TO RANGE FUNCTION IN THIS FUNCTION INSTED OF STOP SIZE THE NUMBER OF EVENLY  SPACED VALUES BETWEEN THE INTERVAL IS SPECIFIED.

```python
K=np.linspace(0,1,50)
```

```python
K
```

```python
n=np.linspace(10,20,5)
```

```python
n
```

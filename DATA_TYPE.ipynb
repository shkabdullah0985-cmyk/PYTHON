# 8)What Is Complex Data Type in Python?

=>


The complex data type (complex) is used to store numbers that have two parts:

. a real part.

. an imaginary part.


---

**Important Rules**

1. The imaginary part uses j, not i.

2. Complex numbers are written without quotes.

3. Python automatically treats numbers with j as complex.

---



```python
''' EXAMPLES COMPLEX DATA TYPE'''
c=3+4j
type(c)
```

```python
a = 3
b= 4j
c=a+b
print(type(c))
```

# 9)HOW TO ACCESS REAL AS WELL AS IMAGINARY PART FROM COMPLEX DATA TYPE ?

=>

 In Python, a complex number has two built-in attributes:

**.real** → gives the real part.

**.imag** → gives the imaginary part.


---



```python
'''EXAMPLE'''
c=5+4j
print(c.real)
print(c.imag)
```

# 15) What Is Dictionary Data Type in Python?
=>

1. A dictionary is a mutable data type used to store data in key-value pairs.
2. Its is surrounded by {}curly braces and seprated by (,).

---

**Dictionary Characteristics**

1. Stores data as key : value.

2. Keys must be unique.

3. Values can be duplicated.

4. Mutable (can be changed).


---



```python
'''EXAMPLES OF DICTIONARY DATA TYPES'''
a={"Python":1991,"C":1972}
print(a)
```

```python
a={}
type(a)
```

```python
a={"steve":1991,"robin":1991,"karan":1987}
print(a)
print(a["steve"])
print(a["robin"])
print(a["karan"])
```

# 16)What Is Set Data Type in Python?
=>

1. A set is a mutable data type used to store multiple unique values in a single variable.
2. Its is surrounded by {}curly braces and seprated by (,) commas.

**Set Characteristics**

```python
a={1,2,3,4,4}
print(a)
type(a)
```

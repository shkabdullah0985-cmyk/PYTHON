# 4) WHAT IS CLASS ?

```python
'''SYNTAX OF CLASS IN PYTHON'''
class CLASSNAME:
  variable=10                    #variables(attributes)

  def method_name(self):         #methods (functions)
    print("THIS IS A METHOD")

```

# 8) What Is Encapsulation in Python?

=> Encapsulation is an OOPS concept in which data (variables) and methods (functions) are wrapped together inside a class, and direct access to data is restricted to protect it.

# 9) Why Use Encapsulation?

=>
1. Protects data from unauthorized access

2. Improves security

3. Makes code easy to maintain

4. Prevents accidental data modification

```python
'''EXAMPLE OF ENCAPLSULATION'''
class HUMAN:
    NAME = "ALI"
    EYES = "BLACK"
    NOSE = "SHARP"
    LEGS = 2
    HANDS = 2
    BRAIN = "ACTIVE"

    def walk():
        print("HUMANS CAN WALK")

    def speak():
        print("HUMANS CAN SPEAK")

    def eat():
        print("HUMANS CAN EAT FOOD")

    def think():
        print("HUMANS CAN THINK")

    def details():
        print(
            HUMAN.NAME,
            HUMAN.EYES,
            HUMAN.NOSE,
            HUMAN.LEGS,
            HUMAN.HANDS,
            HUMAN.BRAIN
        )
print(HUMAN.NAME)
print(HUMAN.EYES)
print(HUMAN.NOSE)
print(HUMAN.LEGS)
print(HUMAN.HANDS)
print(HUMAN.BRAIN)

HUMAN.walk()
HUMAN.speak()
HUMAN.eat()
HUMAN.think()
HUMAN.details()
```

In Python, access modifiers are used to protect data and control visibility of class members.

# Types of Access Modifier:

1. Public	                      
2. Protected (_)
3. Private (__)

**1.Public:-**
 Accessible anywhere.

**2.Protected(_):-**
Accessible within class and subclass.

**3.Private(__):-**
Accessible only inside the class.


---



```python
'''EXAMPLE OF PUBLIC ACCESS MODIFIER'''
class A:
    x = 10

print(A.x)
```

# 12) What Is Inheritance in Python?

=>
 Inheritance is an Object-Oriented Programming concept where one class (child class) acquires the properties and methods of another class (parent class).

# 13) Why We Use Inheritance?
=>
1. Avoids code duplication

2. Improves code reusability

3. Makes programs easier to maintain

4. Supports hierarchical structure

```python
'''SYNTAX OF INHERITANCE'''
#class ChildClass(ParentClass):

```

```python
'''EXAMPLE OF INHERITANCE'''
class Human:
    def walk():
        print("Humans can walk")

class Student(Human):
    def study():
        print("Students can study")

Student.walk()
Student.study()
```

# 14) What is Function Overloading and Function Overriding ?

=>
1. Function Overloading

Function overloading means multiple functions with the same name but different parameters.

2. Function Overriding

Function overriding means a child class provides its own version of a method already defined in the parent class.

```python
'''EXAMPLE OF FUNTION OVERLOADING'''
class Math:
    def add(self, a, b, c=0):
        print(a + b + c)

m = Math()
m.add(2, 3)
m.add(2, 3, 4)
```

```python
'''EXAMPLE OF FUNTION OVERRIDING'''
class Animal:
    def sound(self):
        print("Animal sound")

class Dog(Animal):
    def sound(self):
        print("Dog barks")

a = Animal()
d = Dog()

a.sound()
d.sound()
```

# 15) What Is Polymorphism in Python?
=>
 Polymorphism is an Object-Oriented Programming concept where the same method or function name behaves differently depending on the object or data type.

# 16) Why We Use Polymorphism?

1. Makes code flexible

2. Improves readability

3. Supports method reuse

4. Helps in extending programs

```python
'''EXAMPLE OF POLYMORPHISM'''
class Animal:
    def sound(self):
        print("Animal makes a sound")
class Dog(Animal):
    def sound(self):
        print("Dog barks")

class Cat(Animal):
    def sound(self):
        print("Cat meows")
a = Animal()
d = Dog()
c = Cat()

a.sound()
d.sound()
c.sound()
```

# Types of Polymorphism in Python

1. Function Polymorphism (Built-in)
2.  Method Polymorphism (Inheritance)
3.  Operator Polymorphism


**1)FUNCTION POLYMORPHISM**

=>
   Function polymorphism means the same function name works with different data types and performs different operations depending on the input.

```python
'''EXAMPLE OF FUNCTION POLYMORPHISM'''
print(len("Python"))        # string
print(len([1, 2, 3, 4]))    # list
print(len((10, 20)))        # tuple
```

```python
'''EXAMPLE OF FUNTION POLYMORPHISM USING CLASS AND OBJECTS'''
class Student:
    def info(self):
        return "Student studies"

class Teacher:
    def info(self):
        return "Teacher teaches"

class Doctor:
    def info(self):
        return "Doctor treats patients"
s = Student()
t = Teacher()
d = Doctor()

def show_details(person):
    print(person.info())
show_details(s)
show_details(t)
show_details(d)
```

**2) METHOD POLYMORPHISM**

=>
     Method polymorphism is an Object-Oriented Programming concept where methods with the same name behave differently in different classes.


```python
'''EXAMPLE OF METHOD POLYMORPHISM'''
class Animal:
    def sound(self):
        print("Animal makes a sound")

class Dog(Animal):
    def sound(self):
        print("Dog barks")

class Cat(Animal):
    def sound(self):
        print("Cat meows")
a = Animal()
d = Dog()
c = Cat()

a.sound()
d.sound()
c.sound()

```

**3) OPERATOR POLYMORPHISM**

=>  Operator polymorphism means the same operator performs different operations depending on the data types of the operands.

```python
'''EXAMPLE OF OPERATOR POLYMORPHISM'''
print(10 + 5)
print("Py" + "thon")
print([1, 2] + [3, 4])
```

# 17) WHAT IS ABSTRACTION ?
=>
Abstraction is an Object-Oriented Programming concept that means hiding internal implementation details and showing only what is necessary to the user

```python
'''EXAMPLE OF ABSTRACTION'''
class A:
    x = 10

print(A.x)                         #PUBLIC
class B:
    _y = 20
print(B._y)                        #PROTECTED
class C:
    __z = 30
print(C.__z)                        #PRIVATE
```

# 18) What Is a Constructor in Python?

A constructor is a special method in a class that runs automatically when an object is created.It is mainly used to initialize (set) values for object variables.In Python, the constructor is named

**__init __()**


```python
'''EXAMPLE 1  OF CONSTRUCTOR'''
class Student:
    def __init__(self):
        print("Object is created")

s = Student()
```

```python
'''EXAMPLE 2 OF CONSTRUCTOR'''
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def show(self):
        print(self.name, self.marks)

s1 = Student("Ali", 90)
s1.show()
```

# 19) Why We Use a Constructor ?

=>
1. To initialize object data

2. To give values at object creation

3. To reduce extra code

4. To ensure object starts in a valid state

---



**Important Points**

1. Constructor is called automatically

2. It runs only once per object

3. A class can have only one constructor in Python

4. If not defined, Python provides a default constructor

20) What Is a Destructor in Python?

A destructor is a special method in a class that is called automatically when an object is destroyed or deleted.It is mainly used to release resources such as memory, files, or database connections.In Python, the destructor is named

**__del __()**


```python
'''EXAMPLE OF DESTRUCTOR'''
class Student:
    def __init__(self):
        print("Object created")

    def __del__(self):
        print("Object destroyed")

s = Student()
del s
```

# 21) Why We Use a Destructor ?
=>

1. To close files

2. To release memory

3. To clean up resources

4. To perform final actions before object removal


---


**Important Points to Remember**

1. Destructor is called automatically

2. Python uses garbage collection

3. Destructor execution time is not guaranteed

 4. __del __() may not run immediately

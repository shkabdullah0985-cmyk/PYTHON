#1)WHAT IS NUMPY ?

NUMPY (NUMERICAL PYTHON) IS A POWERFUL PYTHON LIBRARY USED FOR NUMERICAL AND SCIENTIFIC COMPUTING.
IT PROVIDES SUPPORT FOR ARRAYS, MATRICES, AND MANY MATHEMATICAL FUNCTIONS TO PERFORM FAST CALCULATIONS.

NUMPY IS THE FOUNDATION OF DATA SCIENCE, MACHINE LEARNING, AND SCIENTIFIC COMPUTING IN PYTHON BECAUSE IT MAKES WORKING WITH LARGE AMOUNTS OF DATA EASY AND EFFICIENT.

```python
import numpy as np
```

#2) WHAT IS ARRAY() ?
ARRAY() IS A FUNCTION IN THE NUMPY LIBRARY THAT IS USED TO CREATE AN ARRAY FROM A LIST, TUPLE, OR OTHER SEQUENCE.
IT CONVERTS NORMAL PYTHON DATA STRUCTURES INTO NUMPY ARRAYS WHICH ARE FASTER AND MORE POWERFUL FOR NUMERICAL CALCULATIONS.

```python
np.array([[1,2,3],[4,5,6]])
```

```python
a=np.array([[1,2,3],[4,5,6]])
```

#3)WHAT IS shape() IN NUMPY ?

shape IS AN ATTRIBUTE (NOT A FUNCTION) IN NUMPY THAT TELLS THE SIZE OR DIMENSIONS OF AN ARRAY.

IT SHOWS HOW MANY ROWS AND COLUMNS (OR LEVELS) ARE PRESENT IN THE ARRAY.
IN SIMPLE WORDS, IT DESCRIBES THE STRUCTURE OF THE ARRAY.

```python
a.shape
```

#4)WHAT IS size() IN NUMPY

size IS AN ATTRIBUTE IN NUMPY THAT RETURNS THE TOTAL NUMBER OF ELEMENTS PRESENT IN AN ARRAY — REGARDLESS OF ITS DIMENSIONS.

```python
a.size
```

# 7)WHAT IS CONCATENATE() IN NUMPY ?

CONCATENATE() IS A FUNCTION IN NUMPY THAT IS USED TO JOIN (COMBINE) TWO OR MORE ARRAYS INTO A SINGLE ARRAY.

IT PUTS ARRAYS END TO END ALONG A SPECIFIED AXIS —
LIKE JOINING STRINGS OR SENTENCES TOGETHER, BUT HERE IT JOINS NUMPY ARRAYS.

```python
z=np.concatenate((x,a))
```

```python
print(z)
```

# 8)WHAT IS HSTACK() IN NUMPY ?

HSTACK() IS A FUNCTION IN NUMPY THAT IS USED TO JOIN (STACK) ARRAYS HORIZONTALLY,
MEANING — IT COMBINES ARRAYS SIDE BY SIDE (COLUMN-WISE).

THE LETTER ‘H’ IN HSTACK STANDS FOR HORIZONTAL.

```python
m=np.hstack((x,a))
```

```python
print(m)
```

```python
q=np.array([1,2,3,4,5,6])
```

# 9)WHAT IS VSTACK() IN NUMPY ?

VSTACK() IS A FUNCTION IN NUMPY THAT IS USED TO JOIN (STACK) ARRAYS VERTICALLY,
MEANING — IT COMBINES ARRAYS ONE ON TOP OF ANOTHER (ROW-WISE).

THE LETTER ‘V’ IN VSTACK STANDS FOR VERTICAL.

```python
np.vstack(q)
```

#10)WHAT IS APPEND() IN NUMPY ?

APPEND() IS A FUNCTION IN NUMPY THAT IS USED TO ADD (APPEND) NEW VALUES OR ARRAYS TO AN EXISTING ARRAY AND RETURN A NEW ARRAY.

IT DOES NOT CHANGE THE ORIGINAL ARRAY — IT CREATES A NEW ONE WITH THE ADDED ELEMENTS.

```python
ap=np.append(ap,[7,8,9])
```

```python
print(ap)
```

```python
mn=np.array([2,3,4,5])
```

#11)WHAT IS INSERT() IN NUMPY ?

INSERT() IS A FUNCTION IN NUMPY THAT IS USED TO INSERT (ADD) VALUES INTO AN ARRAY AT A SPECIFIC POSITION OR INDEX.

IT CREATES A NEW ARRAY WITH THE INSERTED VALUES —
THE ORIGINAL ARRAY DOES NOT CHANGE.

```python
mn=np.insert(mn,0,1)
```

```python
print(mn)
```

```python
mn=np.insert(mn,0,7)
```

```python
print(mn)
```

#12)WHAT IS DELETE() IN NUMPY ?

DELETE() IS A FUNCTION IN NUMPY THAT IS USED TO REMOVE (DELETE) ELEMENTS, ROWS, OR COLUMNS FROM AN EXISTING ARRAY.

IT DOES NOT CHANGE THE ORIGINAL ARRAY —
INSTEAD, IT RETURNS A NEW ARRAY WITH THE SPECIFIED ELEMENTS DELETED.

```python
mn=np.delete(mn,0)
```

```python
print(mn)
```

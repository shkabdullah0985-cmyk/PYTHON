# 5) What Is a Local Variable?
=>

A local variable is a variable that is defined inside a function and can be used only within that function.

---
**Key Points**

1. Defined inside a function.

2. Scope is limited to that function.

3. Destroyed after function execution.

---




```python
'''EXAMPLES OF LOCAL VARIABLE'''
def show():
    x = 10   # local variable
    print(x)

show()
```

```python
def show():
    t = 10    # Trying to access the local variable outside the class.

print(t)   #  Error Reason: t is local to the function we cannot use local variable outside the function.
```

# 6) What Is a Global Variable ?
=>

A global variable is a variable that is defined outside all functions and can be accessed anywhere in the program.

---
**Important Points**

1. Defined outside functions.

2. Accessible inside and outside functions.

3. global keyword required to modify it inside a function.

---




```python
'''EXAMPLES OF GLOBAL VARIABLE'''
x1 = 10   # global variable

def show():
    print(x1)

show()
```

```python
x1 = 5

def display():
    print(x1)         #Accessing gloal variable inside the function.

display()
```

# 7) How to Modify a Global Variable in Python ?
=>

To modify a global variable inside a function, you must use the **global** keyword.

---
**Important Points**

1. Defined outside functions.

2. Accessible inside and outside functions.

3. global keyword required to modify it inside a function.

---




```python
'''EXAMPLE'''
x = 10

def change():
    global x     # Modifying Global Variable Inside Function
    x = 20

change()
print(x)
```

# 8)What Happens Without global in Python ?
=>


When you do not use the global keyword, Python treats the variable inside the function as a local variable, even if a global variable with the same name exists.

---



```python
'''EXAMPLE'''
x = 10   # global variable

def change():
    x = 20   # local variable

change()
print(x)
```

```python
''' Why This Happens ?
=>

1. x = 20 creates a new local variable.

2. The global variable x is not modified.

3. Local variable exists only inside the function.'''
```

# 9) What Is an Instance Variable?
=>

1. An instance variable is a variable that belongs to a specific object (instance) of a class.
2. Each object has its own copy of an instance variable.

---





```python
'''EXAMPLE OF INSTANCE VARIABLE USING CONSTRUCTOR'''
class Student:
    def __init__(self, name):
        self.name = name   # instance variable

s1 = Student("Ali")
s2 = Student("Ahmed")

print(s1.name)
print(s2.name)
```

```python
'''Explanation

1. self.name is an instance variable.

2. s1 and s2 have different values.

3. Changing one object does not affect the other.'''
```

```python
'''EXAMPLE OF INSTANCE VARIABLE WITHOUT USING CONSTRUCTOR'''
class Student:
    def show(self):
        print(self.name)

s1 = Student()
s2 = Student()

s1.name = "Ali"      # instance variable
s2.name = "Ahmed"   # instance variable

s1.show()
s2.show()
```

# 10)What Is a Class Variable ?
=>

1. A class variable is a variable that belongs to the class itself, not to individual objects.
2. It is shared by all objects created from that class.

---



```python
class Student:
    school = "ABC School"   # class variable

s1 = Student()
s2 = Student()

print(s1.school)
print(s2.school)
```

# 11)Difference Between Class Variable and Instance Variable ?

=>
| Feature              | Class Variable                            | Instance Variable                         |
| -------------------- | ----------------------------------------- | ----------------------------------------- |
| Definition           | Variable that belongs to the **class**    | Variable that belongs to an **object**    |
| Where defined        | Inside the class, outside methods         | Inside methods using `self` or via object |
| Ownership            | Shared by all objects                     | Separate for each object                  |
| Memory               | Created once                              | Created per object                        |
| Access inside class  | `ClassName.variable`                      | `self.variable`                           |
| Access outside class | `ClassName.variable` or `object.variable` | `object.variable`                         |
| Value change effect  | Affects all objects                       | Affects only that object                  |


---



```python
'''EXAMPLE OF CLASS VARIABLE'''
class Student:
    school = "ABC School"   # class variable

s1 = Student()
s2 = Student()

print(s1.school)
print(s2.school)
```

```python
'''EXAMPLE OF INSTANCE VARIABLE'''
class Student:
    def set_name(self, name):
        self.name = name   # instance variable

s1 = Student()
s2 = Student()

s1.set_name("Ali")
s2.set_name("Ahmed")

print(s1.name)
print(s2.name)
```

# 12)What Is Variable Scope?
=>

Variable scope defines where a variable can be accessed or used in a program.

---



# 13)Types of Variable Scope in Python ?
=>
1. LOCAL SCOPE.
2. GLOBAL SCOPE.
3. ENCLOSING SCOPE.
4. BUILT IN SCOPE.

---



# 14)What Is Local Scope?
=>


Local scope refers to variables that are defined inside a function and can be used only within that function.

---



```python
'''EXAMPLES  OF  LOCAL SCOPE'''
def show():
    x = 10   # local variable
    print(x)

show()
```

```python
def show():
    t = 10    # Trying to access the local variable outside the class.

print(t)   #  Error Reason: t is local to the function we cannot use local variable outside the function.
```

# 15)What Is Global Scope?
=>

Global scope refers to variables that are defined outside all functions and can be accessed anywhere in the program.

---



```python
'''EXAMPLES OF GLOBAL SCOPE'''
x = 10   # global variable

def show():
    print(x)

show()
print(x)
```

# 16) What Is Enclosing Scope?
=>

Enclosing scope refers to variables that are defined in an outer function and are accessible inside an inner (nested) function.

---



```python
'''EXAMPLE OF ENCLOSING SCOPE'''
def outer():
    x = 10   # enclosing variable

    def inner():
        print(x)   # accessed from enclosing scope

    inner()

outer()
```

```python
'''Why It Is Called Enclosing

1. x is not local to inner().

2. x is not global.

3. It is enclosed by the outer function.'''
```

# 17)How to Modify an Enclosing Variable in Python ?
=>


To modify an enclosing (outer function) variable from an inner function, you must use the **nonlocal** keyword.

---



```python
'''EXAMPLE'''
def outer():
    x = 10   # enclosing variable

    def inner():
        nonlocal x
        x = 20   # modify enclosing variable

    inner()
    print(x)

outer()
```

# 18)What Happens Without nonlocal in Python ?
=>


When you do not use the nonlocal keyword, Python treats the variable inside the inner function as a local variable, not as an enclosing variable.

---



```python
def outer():
    x = 10

    def inner():
        x = 20   # local variable

    inner()
    print(x)

outer()

```

```python
'''Why This Happens ?

1. x = 20 creates a new local variable inside inner().

2. The enclosing variable x in outer() is not modified.'''
```

# 19)What Is Built-in Scope?
=>


1. Built-in scope refers to the names that are always available in Python, without defining them yourself.

2. These include built-in functions, keywords, and exceptions provided by Python.

---



```python
'''EXAMPLES OF BUILT- IN - SCOPE'''
x = [1, 2, 3, 4]  # (len()) Built-in Function.
print(len(x))     #But here len() is from built-in scope.
```

```python
print(max(10, 20, 30))   # (min() and max())Built-in Function.
print(min(5, 8, 2))      #  But here max() and min() is form built-in scope.
```

```python
a = 10                 #type() is Built- in Function.
print(type(a))         #type() is from built-in scope
```

# 20)Difference Between Local, Global, Enclosing, and Built-in Scope ?
=>
| Scope Type    | Where Variable Is Defined        | Where It Can Be Used            | Example                                 |
| ------------- | -------------------------------- | ------------------------------- | --------------------------------------- |
| **Local**     | Inside a function                | Only inside that function       | `x = 10` inside a function              |
| **Enclosing** | In an outer (enclosing) function | Inside inner (nested) functions | Variable in `outer()` used in `inner()` |
| **Global**    | Outside all functions            | Anywhere in the program         | `x = 10` at top level                   |
| **Built-in**  | Provided by Python               | Available everywhere            | `len()`, `print()`                      |

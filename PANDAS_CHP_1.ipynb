#***1)WHAT IS PANDAS  ?***

PANDAS IS A POWERFUL PYTHON LIBRARY USED FOR DATA ANALYSIS, DATA MANIPULATION, AND DATA CLEANING.
IT PROVIDES EASY-TO-USE DATA STRUCTURES AND TOOLS TO WORK WITH LABELED OR TABULAR DATA (LIKE EXCEL SHEETS OR DATABASE TABLES).


---



**MEANING:**

THE WORD “PANDAS” COMES FROM “PYTHON DATA ANALYSIS LIBRARY.”

IT IS BUILT ON TOP OF NUMPY, WHICH MAKES IT VERY FAST AND EFFICIENT.



---


**MAIN FEATURES OF PANDAS:**

EASY DATA CLEANING AND HANDLING.

SUPPORTS IMPORTING DATA FROM CSV, EXCEL, SQL, JSON, ETC.

POWERFUL OPERATIONS LIKE FILTERING, GROUPING, SORTING, AND MERGING.

BUILT ON TOP OF NUMPY, MAKING IT FAST AND EFFICIENT.

IDEAL FOR DATA SCIENCE AND MACHINE LEARNING PROJECTS.


---



# ***2)WHY WE HAVE TO IMPORT NUMPY BEFORE PANDAS ?***
WE IMPORT NUMPY FIRST BECAUSE PANDAS IS BUILT ON TOP OF NUMPY.
NUMPY PROVIDES THE CORE DATA STRUCTURE (ARRAYS) THAT PANDAS USES INTERNALLY TO STORE AND MANAGE DATA.


---

# DETAILED EXPLANATION:-

PANDAS DEPENDS ON NUMPY

PANDAS USES NUMPY ARRAYS UNDERNEATH ITS OWN DATA STRUCTURES (LIKE SERIES AND DATAFRAME).

ALL NUMERICAL OPERATIONS INSIDE PANDAS (LIKE MEAN, SUM, SORT, ETC.) USE NUMPY FUNCTIONS.

# `SO, WITHOUT NUMPY, PANDAS CANNOT WORK PROPERLY`.


---
NUMPY HANDLES MATHEMATICAL OPERATIONS

PANDAS IS DESIGNED FOR DATA ANALYSIS, NOT FOR LOW-LEVEL MATH OPERATIONS.

NUMPY PROVIDES THOSE MATHEMATICAL AND ARRAY FUNCTIONS (LIKE np.sum(), np.mean(), np.array()), WHICH PANDAS USES INTERNALLY.

DATA STRUCTURE CONNECTION

A PANDAS SERIES IS BASICALLY A WRAPPER AROUND A NUMPY ARRAY WITH ADDED LABELS (INDEX).

A PANDAS DATAFRAME IS A COLLECTION OF MULTIPLE SERIES OBJECTS THAT ALL USE NUMPY ARRAYS UNDER THE HOOD.

---
PANDAS IMPORTS NUMPY AUTOMATICALLY

WHEN YOU import pandas, IT INTERNALLY IMPORTS NUMPY TOO.

BUT, SOMETIMES DEVELOPERS IMPORT NUMPY FIRST TO USE BOTH TOGETHER FOR ADVANCED CALCULATIONS.

---
#  `CONCLUSION:-`

WE IMPORT NUMPY BEFORE PANDAS BECAUSE
PANDAS USES NUMPY’S FUNCTIONS, ARRAYS, AND OPERATIONS INTERNALLY TO WORK WITH DATA EFFICIENTLY.

# ***4)WHY WE IMPORT WARNINGS IN PYTHON***

WARNINGS text IS A BUILT-IN PYTHON MODULE THAT IS USED TO CONTROL, IGNORE, OR MANAGE WARNING MESSAGES THAT APPEAR WHEN A PROGRAM RUNS.

SOMETIMES, WHEN YOU RUN CODE, PYTHON SHOWS WARNINGS (NOT ERRORS) — THESE WARNINGS DO NOT STOP THE PROGRAM, BUT THEY CAN MAKE THE OUTPUT LOOK MESSY OR CONFUSING.

TO HANDLE THIS, WE IMPORT THE WARNINGS MODULE.

---

# WHY WE USE IT ?

🛑 TO HIDE UNNECESSARY WARNING MESSAGES

SOMETIMES, LIBRARIES LIKE NUMPY, PANDAS, OR SCIKIT-LEARN SHOW WARNINGS ABOUT SMALL ISSUES (E.G., DEPRECATED FUNCTIONS).

THESE WARNINGS DO NOT AFFECT THE CODE RESULT, BUT THEY MAKE THE OUTPUT LOOK CLUTTERED.

**TO STOP THEM, WE USE:**


# **`warnings.filterwarnings('ignore')`**


---
**WARNINGS MODULE IS USED TO MANAGE, IGNORE, OR DISPLAY WARNING MESSAGES IN PYTHON PROGRAMS.**

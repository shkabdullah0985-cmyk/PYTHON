#5)WHAT IS BROADCASTING IN NUMPY ARRAYS ?
BROADCASTING ALLOWS NUMPY TO DO OPERATIONS BETWEEN ARRAYS OF DIFFERENT SIZES BY “STRETCHING” THE SMALLER ONE TO FIT THE BIGGER ONE.

```python
M= np.arange(1,7)
```

```python
M[2]=55
```

```python
print(M)
```

```python
M[ : :2]=15
```

```python
print(M)
```

```python
M[1::2]=0
```

```python
print(M)
```

```python
N=np.arange(1,21).reshape(4,5)
```

```python
N
```

```python
N[::4]=55
```

```python
N
```

```python
N[::1,3:4:1]=0.25
```

```python
N
```
